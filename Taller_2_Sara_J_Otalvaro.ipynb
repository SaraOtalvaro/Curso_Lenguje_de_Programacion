#**Gestión de Portafolios Financieros**

## Datos de inversión inicial##
**Clientes:** Sofía Martínez, Luis Torres, Andrea Castro, Miguel Rivas, Juana Posada

**Montos invertidos:** 14000, 54000, 10000, 32000, 26000

**Rendimientos anuales (%):** 5.6, 3.2, 5.0, 4.8, 1.7

##**Declaración de listas y variables: Registro de clientes**

```python
# Declaración de listas
clientes = ["Sofía Martínez", "Luis Torres", "Andrea Castro", "Miguel Rivas", "Juana Posada"]
montos_invertidos = [14000, 54000, 10000, 32000, 26000]
rendimiento = [5.6, 3.2, 5.0, 4.8, 1.7]

print("Clientes:", clientes)
print("Inversion:", montos_invertidos)
print("Rendimiento:", rendimiento)
```

##**Almacenar información: Tuplas**

```python
#creación de la tupla con los datos de cada cliente
cliente_1 = (clientes[0], montos_invertidos[0], rendimiento[0])
cliente_2 = (clientes[1], montos_invertidos[1], rendimiento[1])
cliente_3 = (clientes[2], montos_invertidos[2], rendimiento[2])
cliente_4 = (clientes[3], montos_invertidos[3], rendimiento[3])
cliente_5 = (clientes[4], montos_invertidos[4], rendimiento[4])

print(cliente_1)
print(cliente_2)
print(cliente_3)
print(cliente_4)
print(cliente_5)
```

##**Diccionario y condicionales**

```python
#Diccionario portafolio
portafolio = {
    "Sofía Martínez": { "Inversión": 14000, "Rendimiento": 5.6 },
    "Luis Torres": { "Inversión": 54000, "Rendimiento": 3.2 },
    "Andrea Castro": { "Inversión": 10000, "Rendimiento": 5.0 },
    "Miguel Rivas": { "Inversión": 32000, "Rendimiento": 4.8 },
    "Juana Posada": { "Inversión": 26000, "Rendimiento": 1.7 }
}

print("Portafolio:", portafolio)
```

##**Rendimiento de clientes**

```python
# Identificar clientes con Rendimiento menor a 5%
print("Clientes con Rendimiento menor al 5%:")
for cliente, datos in portafolio.items():
    if datos["Rendimiento"] < 5:
        print(f"{cliente} tiene un Rendimiento de {datos['Rendimiento']}%")
```

##**Cálculos con Ciclos: Cálculo de rendimientos**


```python
#Cálculo de rendimientos
rendimiento_total = 0

for cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["Inversión"] * datos["Rendimiento"]) / 100
    rendimiento_total += rendimiento_cliente

print(f"El rendimiento total del portafolio es: ${rendimiento_total:.2f}")
```

##**Cálculos con Ciclos: Cliente con mayor rendimiento**

```python
mayor_rendimiento = 0
cliente_mayor = ""

for cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["Inversión"] * datos["Rendimiento"]) / 100
    if rendimiento_cliente > mayor_rendimiento:
        mayor_rendimiento = rendimiento_cliente
        cliente_mayor = cliente

print(f"El cliente con el mayor rendimiento absoluto es {cliente_mayor} con un rendimiento de ${mayor_rendimiento:.2f}")
```

##**Modificaciones del portafolio**

```python
# Actualizar el monto invertido de "Luis Torres" a 60,000
portafolio["Luis Torres"]["Inversión"] = 60000
print("Monto actualizado de Luis Torres:", portafolio["Luis Torres"])
```

```python
# Añadir un nuevo cliente: "Giselle García"
portafolio["Giselle García"] = {"Inversión": 22000, "Rendimiento": 6.8}
print("Nuevo cliente agregado:", portafolio["Giselle García"])
```

```python
# Remover a un cliente: "Juana Posada"
del portafolio["Juana Posada"]

print("Cliente 'Juana Posada' eliminado. Clientes actuales:")
for cliente in portafolio:
    print("-", cliente)
```
